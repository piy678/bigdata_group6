In diesem Schritt speichern wir die bereinigten Datensätze aus unseren drei Quellen (CSV-Dateien) dauerhaft ab.  
Verwendet wird Apache Spark, um die Daten ins Parquet-Format zu überführen, das effizient und für spätere Analysen optimiert ist.

## 1. Spark Session starten

Zuerst wird eine Spark Session gestartet, die die Verarbeitung der Daten ermöglicht.

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Data Storage") \
    .getOrCreate()
```

## 2. CSV-Dateien laden

Wir lesen die bereinigten CSV-Dateien aus dem Ordner `data/processed/` ein.  
Sie stammen aus unseren drei Datenquellen:

- `spotify_cleaned.csv` → aus einer Datei (Top 1000 Spotify-Songs)
- `api_cleaned.csv` → aus einer REST API (Live-Musikdaten)
- `webscraping_cleaned.csv` → aus Web Scraping (z. B. Charts, Empfehlungen)

```python
df_spotify = spark.read.option("header", True).csv("../../../data/processed/spotify_cleaned.csv")
df_api = spark.read.option("header", True).csv("../../../data/processed/api_cleaned.csv")
df_web = spark.read.option("header", True).csv("../../../data/processed/webscraping_cleaned.csv")
```

## 3. Speicherung als Parquet

Die geladenen DataFrames werden nun als Parquet-Dateien im `output/` Ordner gespeichert.  
Parquet ist ein spaltenbasiertes, komprimiertes Format – ideal für Big Data-Analysen mit Spark oder andere Tools.

```python
df_spotify.write.mode("overwrite").parquet("../../../output/spotify_data.parquet")
df_api.write.mode("overwrite").parquet("../../../output/api_data.parquet")
df_web.write.mode("overwrite").parquet("../../../output/webscraping_data.parquet")

print("✅ Daten erfolgreich als Parquet gespeichert.")
```

## 4. Vorschau der gespeicherten Daten

Wir laden die gespeicherten Parquet-Dateien testweise ein und zeigen einige Zeilen zur Kontrolle.

```python
df_spotify.show(5)
df_api.show(5)
df_web.show(5)
```

```python
# API-Daten speichern
df_api.write.mode("overwrite").parquet("../../../output/api_data.parquet")

# Webscraping-Daten speichern
df_web.write.mode("overwrite").parquet("../../../output/webscraping_data.parquet")

print("✅ API und Webscraping Daten gespeichert.")
```

```python
# Lade gespeicherte Parquet-Dateien
df1 = spark.read.parquet("../../../output/spotify_data.parquet")
df2 = spark.read.parquet("../../../output/api_data.parquet")
df3 = spark.read.parquet("../../../output/webscraping_data.parquet")

# Zeige ein paar Zeilen
df1.show(3)
df2.show(3)
df3.show(3)
```

## Fazit

Alle drei Datenquellen wurden erfolgreich eingelesen und als Parquet gespeichert.  
Damit ist der Data Storage abgeschlossen. Die Daten stehen jetzt für Visualisierungen und Analysen zur Verfügung.
